```python
import pandas as pd

df = pd.read_csv(r"C:\Users\catra\Downloads\breathing_2026-10-01_13-02-01.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

plt.title('BPM over time')
plt.plot(df.time_s, df.bpm_live, alpha = 0.5)
plt.plot(df.time_s, df.bpm, alpha = 0.5)
plt.ylabel("BPM")
plt.xlabel("Time (s)")
gca = plt.gca()
gca.spines['top'].set_visible(False)
gca.spines['right'].set_visible(False)
plt.show()
```

```python
plt.plot(df.time_s, df.value, alpha = 0.5)
gca = plt.gca()
gca.spines['top'].set_visible(False)
gca.spines['right'].set_visible(False)
plt.ylabel('Raw respiration val')
plt.xlabel("Time (s)")
plt.show()
```
